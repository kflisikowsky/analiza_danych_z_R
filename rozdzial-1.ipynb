# 1.2. Instalacja i konfiguracja R oraz RStudio

W tej części przygotujesz środowisko do nauki R: zainstalujesz R i RStudio,
sprawdzisz ich działanie oraz utworzysz pierwszy projekt.
Cele i organizację kursu znajdziesz w [podrozdziale wprowadzającym](rozdzial-1-wprowadzenie.ipynb).

## Cele

Po ukończeniu tej części będziesz umieć:

- zainstalować R oraz RStudio,
- sprawdzić, czy R działa poprawnie,
- utworzyć projekt RStudio,
- instalować i ładować pakiety,
- rozpoznać najczęstsze problemy z konfiguracją.


## 1.2.1. R i RStudio - dwie różne rzeczy

**R** jest językiem programowania i silnikiem obliczeniowym. Wykonuje
polecenia, obliczenia i funkcje.

**RStudio** jest środowiskiem programistycznym (IDE), które ułatwia pracę z
R. Udostępnia między innymi edytor skryptów, konsolę, podgląd wykresów,
menedżer pakietów i obsługę projektów.

Najpierw instalujemy R, a dopiero potem RStudio. RStudio korzysta z R, więc
samo RStudio bez instalacji R nie wystarczy.


## 1.2.2. Instalacja R

Pobierz instalator R z oficjalnej strony projektu CRAN: https://cran.r-project.org/ 

Wybierz system operacyjny:

- **Windows**: `Download R for Windows` → `base` → instalator dla Windows,
- **macOS**: `Download R for macOS` → pakiet `.pkg` odpowiedni dla systemu,
- **Linux**: wybierz dystrybucję i postępuj zgodnie z instrukcjami CRAN.

Podczas instalacji można pozostawić ustawienia domyślne. Po instalacji warto
uruchomić R z menu systemowego i wykonać proste obliczenie:


````r
2 + 2
````

## 1.2.3. Instalacja RStudio

RStudio Desktop pobierzesz ze strony Posit:

<https://posit.co/download/rstudio-desktop/>

Wybierz bezpłatną wersję **RStudio Desktop Open Source License** dla swojego
systemu. Po instalacji uruchom RStudio i sprawdź, czy w panelu **Console**
pojawia się wersja R.

Jeśli RStudio nie znajduje R automatycznie, wybierz interpreter w:

`Tools` → `Global Options` → `General` → `R version`.


## 1.2.4. Pierwszy test instalacji

Poniższa komórka pokazuje wersję R, katalog instalacyjny oraz podstawowe
informacje o systemie. Jeśli wykonuje się bez błędu, R jest gotowy do pracy.


````r
R.version.string
R.home()
Sys.info()[c("sysname", "release", "machine")]
````

## 1.2.5. Projekt w RStudio

Projekt porządkuje skrypty, dane i wyniki w jednym katalogu. Utwórz go przez:

`File` → `New Project` → `New Directory` → `New Project`.

Wybierz prostą nazwę bez spacji i polskich znaków, na przykład
`analiza_danych`. RStudio utworzy plik projektu z rozszerzeniem `.Rproj`.

Pracuj wewnątrz projektu zamiast ręcznie zmieniać katalog roboczy funkcją
`setwd()`. Dzięki temu ścieżki względne pozostają przenośne.


````r
getwd()
list.files()

# Przykład obiektu, który możemy utworzyć w projekcie
powitanie <- "Witaj w RStudio!"
powitanie
````

## 1.2.6. Pakiety R

Pakiet to zestaw funkcji, danych i dokumentacji. Pakiet instalujemy tylko raz,
a w każdej nowej sesji R ładujemy go funkcją `library()`.

Przykład instalacji pakietu `tidyverse`:

```r
install.packages("tidyverse")
library(tidyverse)
```

Instalowanie pakietów wykonuj w konsoli, a wywołania `library()` możesz
zapisywać na początku skryptu. Pakiety pobierane z CRAN powinny pochodzić z
zaufanego źródła.


## 1.2.7. Sprawdzenie pakietów i biblioteki

Nie instalujemy pakietu przy każdym uruchomieniu skryptu. Możemy natomiast
sprawdzić, czy jest dostępny:


````r
pakiet <- "stats"

if (requireNamespace(pakiet, quietly = TRUE)) {
  message("Pakiet jest dostępny: ", pakiet)
  packageVersion(pakiet)
} else {
  message("Pakiet nie jest zainstalowany: ", pakiet)
}
````

## 1.2.8. Codzienna praca w RStudio

Najważniejsze panele RStudio:

- **Source** - skrypty i pliki notebooków,
- **Console** - wykonywanie poleceń R,
- **Environment** - obiekty utworzone w sesji,
- **Files / Plots / Packages / Help** - pliki, wykresy, pakiety i pomoc.

Dobry pierwszy skrypt może zaczynać się tak:

```r
# Analiza danych - rozdział 1
rm(list = ls())

cat("Sesja R działa poprawnie.\n")
```

### Najczęstsze problemy

- **RStudio nie widzi R**: sprawdź `Tools` → `Global Options` → `R version`.
- **Pakiet się nie instaluje**: sprawdź połączenie z internetem i uprawnienia.
- **Nie działa polska czcionka**: zapisz plik jako UTF-8.
- **Nie znaleziono pliku**: otwórz właściwy projekt `.Rproj` i sprawdź `getwd()`.
