# Rozdział 1. Wstęp do języka R

R jest językiem programowania i środowiskiem obliczeniowym używanym do analizy danych, statystyki i wizualizacji. W tym rozdziale poznasz podstawy składni R przed rozpoczęciem właściwej pracy z danymi.

## Obliczenia i przypisania

Komentarze zaczynają się od `#`. Wartości przypisujemy do obiektów operatorem `<-`.

```python
# Proste obliczenie
podstawa <- 8
wykladnik <- 2
pole <- podstawa^wykladnik
pole
```

## Typy danych

R obsługuje między innymi liczby, teksty, wartości logiczne oraz brakujące wartości `NA`.

```python
liczba <- 3.14
tekst <- "język R"
logiczna <- TRUE
brak_danych <- NA

typeof(liczba)
typeof(tekst)
typeof(logiczna)
```

## Wektory i indeksowanie

Wektor tworzymy funkcją `c()`. Indeksowanie w R zaczyna się od `1`.

```python
temperatury <- c(18, 21, 19, 23, 20)
temperatury[1]
temperatury[2:4]
temperatury[temperatury > 20]
```

## Ramki danych i listy

```python
dane <- data.frame(
  imie = c("Anna", "Bartek", "Celina"),
  wiek = c(21, 25, 23),
  student = c(TRUE, FALSE, TRUE)
)

dane
str(dane)

profil <- list(imie = "Anna", wiek = 21)
profil$imie
```

## Funkcje, warunki i pętle

```python
powitaj <- function(imie) {
  paste("Witaj,", imie, "!")
}

powitaj("R")

wynik <- 78
if (wynik >= 60) {
  print("Zaliczone")
} else {
  print("Nie zaliczone")
}

for (liczba in 1:5) {
  print(liczba^2)
}
```
