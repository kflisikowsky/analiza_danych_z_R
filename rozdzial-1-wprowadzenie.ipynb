# 1.1. Wprowadzenie i organizacja kursu

Witaj w kursie **Analiza danych z R**. Jego celem jest zdobycie umiejętności analizowania i prezentowania danych: od sformułowania pytania, przez wybór odpowiedniej metody, aż po wyjaśnienie wyniku. R będzie naszym narzędziem pracy, a statystyka pomoże ocenić, co rzeczywiście wynika z obliczeń.

Ten e-book łączy objaśnienia, przykłady i zadania. Możesz korzystać z niego podczas zajęć oraz wracać do wybranych fragmentów w trakcie samodzielnej pracy. Na początek poznasz zakres kursu, sposób uczenia się i organizację pracy. Następnie przygotujesz środowisko R i RStudio.

> **Informacje o edycji kursu.** Podrozdział opracowano na podstawie prezentacji *AD-W1: Analiza danych z R — wykład, semestr zimowy 2025*. Przytoczone niżej zasady organizacyjne i oceniania opisują tę edycję. Aktualne terminy, punktację, wymagania oraz konsultacje sprawdź w bieżącym kursie na eNauczaniu.


## Czego nauczysz się na kursie?

W materiałach źródłowych cel sformułowano jako nabycie umiejętności analizowania i prezentacji danych. Oznacza to, że po wykonaniu analizy powinieneś umieć odpowiedzieć na trzy pytania: **co badam, dlaczego stosuję tę metodę i jak interpretuję wynik?** Samo otrzymanie liczby lub wykresu nie kończy pracy.

Zakres kursu obejmuje:

- **opracowanie i prezentację materiału statystycznego** — przygotowanie danych oraz przedstawienie ich w czytelnej postaci;
- **populację generalną, próbę i statystykę** — rozróżnienie zbiorowości, o której chcemy wnioskować, od danych, którymi dysponujemy;
- **analizę opisową** — scharakteryzowanie danych i ich zróżnicowania;
- **analizę korelacji** — badanie współwystępowania zmian cech;
- **analizę regresji** — opisywanie zależności za pomocą modeli;
- **modele szeregów czasowych** — analizowanie obserwacji uporządkowanych w czasie.

Ważnym efektem kształcenia jest dobór narzędzi statystycznych do problemu badawczego oraz interpretacja danych i wyników testów. Będziemy więc zwracać uwagę zarówno na obliczenia, jak i na założenia metod oraz sens wyciąganych wniosków.


## Jak pracować z materiałem?

Kurs ma charakter praktyczny. W prezentacji określono go jako „odmatematyzowany”: nacisk położono na rozumienie problemów, korzystanie z programów i interpretację wyników. W e-booku przyjmujemy tę samą zasadę. Wzory i polecenia mają pomagać w analizie konkretnego problemu.

Pracuj w krótkim cyklu:

1. **Przeczytaj objaśnienie przed ćwiczeniem.** Ustal, na jakie pytanie ma odpowiedzieć analiza i co oznaczają używane pojęcia.
2. **Uruchom przykład.** Porównaj wynik z komentarzem w tekście.
3. **Zmień jeden element.** Zmodyfikuj dane lub parametr i sprawdź, jak wpłynęło to na rezultat.
4. **Zapisz interpretację własnymi słowami.** Wyjaśnij znaczenie wyniku w kontekście zadania.
5. **Wykonaj podobne zadanie samodzielnie.** Jeśli pojawi się problem, zanotuj polecenie, komunikat i pytanie do prowadzącego.

Na laboratorium przychodź po zapoznaniu się z teorią dotyczącą danego tematu. Pokazy i wspólne rozwiązywanie zadań są okazją do zadawania pytań. Przepisywanie kodu warto zawsze połączyć z próbą wyjaśnienia, co robi każda jego część.


## E-book, notebook i środowisko R

**Notebook Jupyter** to dokument z komórkami tekstowymi i komórkami kodu. Tekst prowadzi przez zagadnienie, a kod pozwala wykonać obliczenia. W tym kursie kod jest zapisany w języku **R**. Do wykonywania go bezpośrednio w notebooku potrzebne jest jądro R (IRkernel); wybranie jądra Pythona nie pozwoli uruchomić tych przykładów.

Możesz również czytać e-book w przeglądarce i przepisywać lub kopiować przykłady do skryptu w RStudio. W widoku samej książki przyciski uruchamiania kodu mogą nie być dostępne. **R** wykonuje obliczenia, natomiast **RStudio** ułatwia pisanie kodu i przeglądanie wyników. Instalację obu narzędzi omawiają dalsze części rozdziału.

Uruchamiaj komórki kodu po kolei, ponieważ późniejsze przykłady mogą korzystać z wcześniej utworzonych obiektów. Zapisuj własną kopię notebooka wraz z odpowiedziami. Przed zakończeniem pracy sprawdź, czy potrafisz odtworzyć wyniki od początku w nowej sesji R.

W archiwalnej prezentacji wymieniono także Git, GitHub Education i VS Code. Są to dodatkowe elementy warsztatu pracy; ten rozdział prowadzi przez przygotowanie R i RStudio.

## Precyzyjny język opisu danych

Już od pierwszych zajęć dbaj o nazwy i jednostki. **Cyfra** jest znakiem służącym do zapisywania liczb: zapis liczby 23 składa się z cyfr 2 i 3. W opisach danych używaj określenia **liczba** dla obiektów policzalnych, np. „liczba studentów” lub „liczba obserwacji”, a **ilość** dla wielkości takich jak ilość wody czy mąki.

Rozróżniaj również **procenty** i **punkty procentowe**. Wzrost odsetka z 40% do 50% wynosi 10 punktów procentowych. Względem wartości początkowej jest to wzrost o 25%, ponieważ `(50 - 40) / 40 = 0,25`. W raporcie zawsze wskaż, którą zmianę opisujesz.
